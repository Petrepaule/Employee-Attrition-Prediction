# Employee Attrition Prediction Using Machine Learning

# PHASE 1: Data Understanding & Cleaning

## Project Background

Employee attrition is a major challenge for organizations because replacing experienced employees requires significant time, effort, and financial resources. High turnover can disrupt business operations, reduce productivity, lower employee morale, and increase recruitment and training costs.

This project uses the IBM HR Analytics Employee Attrition dataset to explore the factors influencing employee turnover and develop a machine learning model capable of predicting whether an employee is likely to leave the organization. The project begins with understanding the dataset, assessing its quality, and preparing it for exploratory data analysis and predictive modeling.

---

## Project Objectives

The objectives of this phase are to:

- Understand the structure and contents of the dataset.
- Assess the overall quality of the data.
- Identify missing values, duplicates, and inconsistent data types.
- Detect identifier and constant variables.
- Document data cleaning decisions.
- Prepare the dataset for Exploratory Data Analysis (EDA) and Machine Learning.

---

## Dataset Overview

- **Dataset:** IBM HR Analytics Employee Attrition & Performance
- **Prediction Task:** Binary Classification
- **Target Variable:** Attrition
- **Number of Records:** 1,470 Employees
- **Number of Variables:** 35 Features


#### Importing Libraries

```python
import warnings
warnings.filterwarnings('ignore')

import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns

pd.set_option('display.max_columns', None)
pd.set_option('display.float_format', '{:.2f}'.format)

plt.style.use('ggplot')
```

```python
# Loading Dataset and Preview

df = pd.read_csv("WA_Fn-UseC_-HR-Employee-Attrition.csv")
df.head()
df.tail()
df.sample(5)
```

```python
# Dataset Dimensions to Determine the number of observations and variables contained in the dataset.

print(f"Rows: {df.shape[0]}")
print(f"Columns: {df.shape[1]}")
```

```python
# Missing Value Assessment

missing = pd.DataFrame({
    "Missing Values": df.isnull().sum(),
    "Percentage": (df.isnull().mean()*100).round(2)
})

missing
```

This shows that dataset contains no missing values, indicating that all observations are complete and no imputation is required at this stage.

```python
# Duplicate Records

duplicates = df.duplicated().sum()

print(f"Duplicate Records: {duplicates}")
```

```python
# Data Type to determine whether each variable has an appropriate data type.

df.dtypes
```

```python
# Number of Unique Values to identify identifier variables, constant variables, and categorical features that may require encoding during model development.

df.nunique().sort_values()
```

```python
# Constant Features 

constant_columns = [col for col in df.columns if df[col].nunique() == 1]

constant_columns
```

The identified variables contain only a single unique value across all observations.

```python
# Identifier Variables 

df["EmployeeNumber"].nunique()
```

`EmployeeNumber` uniquely identifies each employee and serves as an identifier rather than a predictive feature. It will be excluded during model development.

```python
# Target Variable Inspection

df["Attrition"].value_counts()
```

```python
(df["Attrition"].value_counts(normalize=True) * 100).round(2)
```

The target variable is imbalanced, with significantly more employees remaining in the organization than leaving. This imbalance should be considered during model development and evaluation.

```python
# Data Quality Assessment Summary

summary = pd.DataFrame({
    "Assessment": [
        "Missing Values",
        "Duplicate Records",
        "Constant Columns",
        "Identifier Column",
        "Target Variable"
    ],
    "Result": [
        missing["Missing Values"].sum(),
        duplicates,
        len(constant_columns),
        "EmployeeNumber",
        "Attrition"
    ]
})

summary
```

#### Data Cleaning Decisions

Based on the data quality assessment, the following decisions were made:

- No missing values were detected.
- No duplicate records were identified.
- Three constant variables were identified.
- EmployeeNumber is a unique identifier and will be excluded during feature engineering.
- No datatype inconsistencies were observed.
- The dataset is suitable for Exploratory Data Analysis (EDA).

#### Initial Observations

The dataset is clean, complete, and well-structured for analysis. Most variables are suitable for exploratory analysis, while a few constant and identifier variables have been documented for removal during feature engineering. The target variable exhibits class imbalance, which should be considered during the modeling stage.
